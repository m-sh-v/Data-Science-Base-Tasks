# Часть 2:

- Построим решающую функцию по каждой переменной на основе частот (в качестве порога бинаризации возьмём 0.5).

- Вычислим точность классификации по каждому уникальному значению каждой переменной (на обучающей и тестовой выборке).

- Найдём наиболее информативную переменную (с минимальным числом ошибок).

```python
df = pd.DataFrame(columns=['feature', 'val', 'e_freq', 'p_freq', 'pred'])

for feature in mushrooms.columns:
  if feature == 'class':
    continue
  for value in mushrooms[feature].unique():
    e_freq = mushrooms[
        (mushrooms[feature] == value) & (mushrooms['class'] == 'e')
        ].shape[0] / mushrooms[
            (mushrooms[feature] == value)
        ].shape[0]

    p_freq = mushrooms[
        (mushrooms[feature] == value) & (mushrooms['class'] == 'p')
        ].shape[0] / mushrooms[
            (mushrooms[feature] == value)
        ].shape[0]

    if e_freq > 0.5:
      pred = 'e'
    else:
      pred = 'p'

    df = df._append({'feature':feature, 'val': value, 'e_freq': e_freq, 'p_freq': p_freq,'pred': pred}, ignore_index = True)

for name in df['feature'].unique():
  display(df.loc[df['feature'] == name])
```

```python
all_train = X_train
all_train['class'] = Y_train
display(all_train)
```

```python
all_test = X_test
all_test['class'] = Y_test
display(all_test)
```

```python
for feature in my_categories:


  values_freq = df.loc[df['feature'] == feature].drop(columns=['feature', 'e_freq', 'p_freq'])['val']
  pred_freq = df.loc[df['feature'] == feature].drop(columns=['feature', 'e_freq', 'p_freq'])['pred']
  res = 0
  res_test = 0

  for val, pred in zip(values_freq, pred_freq):
    n = all_train.loc[all_train[feature] == val]['class'].shape[0]
    k = 0
    for cls in all_train.loc[all_train[feature] == val]['class']: # cls = class
      if cls == pred:
        k += 1
    res += k/n

    n_test = all_test.loc[all_test[feature] == val]['class'].shape[0]
    k_test = 0
    for cls in all_test.loc[all_test[feature] == val]['class']:
      if cls == pred:
        k_test += 1
    res_test += k_test/n_test

  print(f'Качаство по признаку {feature} на ОБУЧАЮЩЕЙ выборке = {res/values_freq.shape[0]}')
  print(f'Качаство по признаку {feature} на ТЕСТОВОЙ выборке = {res_test/values_freq.shape[0]}')
```



---



Вывод: по результатам рассчёта качества видно, что наиболее информативным признаком является odor. На обучающей и тестовой выборке он показывает значения accuracy = 0.996 и 0.997

# Часть 4:

Самостоятельно реализуем "наивный" байесовский классификатор, не используя готовый. Сравним полученное решение с библиотечным. Добавим регуляризацию в оценки частот.

```python
data_for_l = X_test_encoded
data_for_l['class'] = Y_test_encoded
```

```python
data_for_l[:3]
```

```python
import numpy as np
```

```python
# числа объектов первого и второго образов
N_1 = list(Y_test).count('e')
N_2 = list(Y_test).count('p')
N = N_1 + N_2

# априорные вероятности
P_1 = N_1/N
P_2 = N_2/N
```

```python
N_1, N_2, N, P_1, P_2
```

```python
def b_j(p, q):
  return np.log(
      (p*(1-q)) /
      (q*(1-p))
  )
```

```python
def b_0(p, q):
  return np.log(
      (1-p)/(1-q)
  )
```

```python
# data = X_test_enc + Y_test_enc

def l_0(data, alpha=0.001):
  result = []
  data = data.reset_index(drop=True)
  for i in range(data.shape[0]):
    b0_j = 0
    sum = 0
    for j in data.columns:
      if j == 'class':
        continue
      val = data.loc[i, j] # x в формуле
      N_x_1 = data.loc[(data[j] == val) & (data['class'] == 0)].shape[0] #e
      N_x_2 = data.loc[(data[j] == val) & (data['class'] == 1)].shape[0] #p

      # регуляризация
      N_x_1 += alpha
      N_x_2 += alpha

      p = N_x_1/(N_1 + alpha)
      q = N_x_2/(N_2 + alpha)

      bj = b_j(p, q)
      b0_j += b_0(p, q)

      sum += bj*val

    b0 = b0_j + np.log(P_1/P_2)
    result.append((sum + b0))

  return result
```

```python
fin = l_0(data_for_l)
```

```python
fin[:10]
```

```python
check = []

for i in fin:
  if i > 0:
    check.append(0) #e
  else:
    check.append(1) #p
```

```python
print(classification_report(Y_test_encoded, check, target_names=['edible', 'poisoned'], digits=4))
```
